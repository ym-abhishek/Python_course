### Datatypes
#### There are 2 types of datatypes, they are: 1. Built-in datatypes 2. User-defined datatypes
1. Built-in datatypes: The datatypes which are already available in python language are called built-in datatypes.
2. User-defined datatypes: The datatypes which are created by the programmers are called User-defined datatypes. 

Built-in datatypes are of 5 types, they are:
1. None Type
2. Numeric Type
3. Sequences
4. Sets
5. Mappings

```python
# None Type: Represents an object that does not contain any value.
a = None
print(a)
```

```python
# Numeric Type: This are related with numbers.
# int type
a = 25
print(a)
```

```python
# float type
b = 0.25
print(b)
```

```python
# complex type
c = 2 + 5j
print(c)
```
