### Comments
#### There are two types of comments in python there are:
1. Single Line Comment 
2. Multi Line Comment

### Single Line Comment
* A single-line comment does not take up memory during execution.
* It is ignored by the Python interpreter.

```python
# Single Line comment
print("This is a single line comment.")
# print this is a single line comment.
```

### Multi Line Comment 
* Multi-line comments do consume storage in the file.
* They are ignored by the Python interpreter and not treated as executable code.
* Doc string ("""..""", '''..''') is used to create multiline comments. 

```python
# Multi Line Comment
print("This is a Multi Line Comment")
print("This is a Multi Line Comment")
print("This is a Multi Line Comment")
""" print("This is a Multi Line Comment")
print("This is a Multi Line Comment")
print("This is a Multi Line Comment") """ 
```
