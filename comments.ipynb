### Comments
#### There are two types of comments in python there are:
1. Single Line Comment 
2. Multi Line Comment

### Single Line Comment
* A single-line comment does not take up memory during execution.
* It is ignored by the Python interpreter.

```python
# Single Line comment
print("This is a single line comment.")
# print this is a single line comment.
```

### Multi Line Comment 
* Multi-line comments do occupy space in the source file.
* The Python interpreter ignores them during execution when they are not assigned to any variable.
* In Python, multi-line comments are not officially recognized as true comments; they are actually multi-line strings.
* When assigned to a variable or placed inside functions/classes, they act as strings known as docstrings.
* Docstrings ("""...""" or '''...''') are mainly used to document code, not just for commenting.

```python
# Multi Line Comment
print("This is a Multi Line Comment")
print("This is a Multi Line Comment")
print("This is a Multi Line Comment")
""" print("This is a Multi Line Comment")
print("This is a Multi Line Comment")
print("This is a Multi Line Comment") """ 
```
